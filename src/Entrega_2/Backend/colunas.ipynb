```python
import pandas as pd

df_player = pd.read_csv(r"data\base_de_dados\base_players.csv", sep=",")
df_lojas = pd.read_csv(r"data\base_de_dados\base_lojas.csv", sep=",")
df_simulacao = pd.read_csv(r"data\base_de_dados\base_simulacao.csv", sep=",")
df_transacoes = pd.read_csv(r"data\base_de_dados\base_transacoes.csv", sep=",")
```
