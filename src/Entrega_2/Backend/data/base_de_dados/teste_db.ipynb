```python
query = "WITH cat_por_player AS (SELECT co.id_player, l.categoria_estabelecimento AS categoria, COUNT(*) AS qtd, ROW_NUMBER() OVER (PARTITION BY co.id_player ORDER BY COUNT(*) DESC, l.categoria_estabelecimento ASC
    ) AS rn
  FROM compra co
  JOIN lojas l ON l.id = co.id_loja
  GROUP BY co.id_player, l.categoria_estabelecimento
)
SELECT
  p.id AS player_id,
  p.celular,
  p.data_nascimento,
  /* SQLite: */ CAST(FLOOR( (julianday('now') - julianday(p.data_nascimento)) / 365.25 ) AS INT) AS idade,
  /* PostgreSQL (use esta linha em vez da de cima):
     -- DATE_PART('year', AGE(CURRENT_DATE, p.data_nascimento))::int AS idade, */
  /* MySQL (use esta linha em vez da de cima):
     -- TIMESTAMPDIFF(YEAR, p.data_nascimento, CURDATE()) AS idade, */
  p.sexo,
  p.cidade_residencial, p.bairro_residencial,
  p.cidade_trabalho,    p.bairro_trabalho,
  p.cidade_escola,      p.bairro_escola,
  cp.categoria AS categoria_frequentada
FROM players p
LEFT JOIN cat_por_player cp
  ON cp.id_player = p.id AND cp.rn = 1;
"
```
