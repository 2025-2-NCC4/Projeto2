## Base Cadastral de Players

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('Base_Cadastral_de_Players.csv', delimiter=';')
print(df.info())
```

```python
nulos = df.isnull().sum() #verifica quantos valores nulos há por coluna
print(nulos[nulos > 0]) #imprime as colunas que tem valores nulos e quantos
```

```python
porcentagem_nulos = df.isnull().mean() * 100
colunas_com_nulos = porcentagem_nulos[porcentagem_nulos > 0]
print(colunas_com_nulos)
```
