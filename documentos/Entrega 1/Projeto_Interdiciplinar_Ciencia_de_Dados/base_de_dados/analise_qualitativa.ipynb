```python
import pandas as pd
from datetime import date
import matplotlib.pyplot as plt
import numpy as np


df_players = pd.read_csv("Base_Cadastral_de_Players.csv", sep=";")
df_transacoes = pd.read_csv("Base_de_Transacoes_e_Cupons_Capturados.csv", sep=";")
df_lojas = pd.read_csv("Base_Massa_de_Teste_com_Lojas_e_Valores.csv", sep=";")
df_simulacao = pd.read_csv("Base_Simulada_-_Pedestres_Av__Paulista.csv", sep=";")
```

```python
def resumo(df):
    return pd.DataFrame({
        "dtype": df.dtypes,
        "na": df.isna().sum(),
        "% preenchido": (df.notna().mean()*100).round(2),
        "n_distintos": df.nunique()
    })
```

```python
df_filtrado = df_simulacao[df_simulacao["possui_app_picmoney"] == "Sim"]

resumo(df_filtrado)
```

```python
lojas_trans = set(df_transacoes["nome_estabelecimento"].unique())
lojas_base  = set(df_lojas["nome_loja"].unique())

lojas_comuns = lojas_trans & lojas_base 
print(len(lojas_trans), len(lojas_base), len(lojas_comuns))
```

```python
resumo_df = resumo(df_players)
resumo_df.to_excel("df_players.xlsx", index=True)

print(resumo_df)
```

```python
resumo_df = resumo(df_transacoes)
resumo_df.to_excel("df_transacoes.xlsx", index=True)

print(resumo_df)
```

```python

resumo_df = resumo(df_lojas)
resumo_df.to_excel("df_lojas.xlsx", index=True)

print(resumo_df)
```

```python
print(resumo(df_simulacao))
resumo_df = resumo(df_simulacao)
resumo_df.to_excel("df_simulacao.xlsx", index=True)

print(resumo_df)
```

```python

df_players = df_players.rename(columns={"celular": "telefone"})
df_transacoes = df_transacoes.rename(columns={"celular": "telefone"})
df_lojas = df_lojas.rename(columns={"numero_celular": "telefone"})
df_simulacao = df_simulacao.rename(columns={"celular": "telefone"})


tel_players = set(df_players["telefone"].dropna().unique())
tel_transacoes = set(df_transacoes["telefone"].dropna().unique())
tel_lojas = set(df_lojas["telefone"].dropna().unique())
tel_simulacao = set(df_simulacao["telefone"].dropna().unique())

comum_todas = tel_players & tel_transacoes & tel_lojas & tel_simulacao

print(f"Telefones em comum entre TODAS as bases: {len(comum_todas)}")

comum_players_transacoes = tel_players & tel_transacoes
print(f"Em comum entre Players e Transações: {len(comum_players_transacoes)}")

comum_players_lojas = tel_players & tel_lojas 
print(f"Em comum entre Players e Lojas: {len(comum_players_lojas)}")

comum_players_simulacao = tel_players & tel_simulacao
print(f"Em comum entre Players e Simulacao: {len(comum_players_simulacao)}")
```
