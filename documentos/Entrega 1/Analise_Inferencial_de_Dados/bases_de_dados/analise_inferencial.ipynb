```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df_transacoes = pd.read_csv("Base_de_Transacoes_e_Cupons_Capturados.csv", sep=";")
```

```python

col1 = "valor_cupom"
col2 = "repasse_picmoney"

x = pd.to_numeric(df_transacoes[col1], errors="coerce")
y = pd.to_numeric(df_transacoes[col2], errors="coerce")

r = x.corr(y, method="pearson")

print(f"Correlação entre {col1} e {col2}: {r:.4f}")
```

```python
mask = x.notna() & y.notna()
xv, yv = x[mask], y[mask]

plt.scatter(xv, yv, alpha=0.1, s=20, label="Dados")

m, b = np.polyfit(xv, yv, 1)
plt.plot(xv, m*xv + b, color="red", linewidth=2, label="Linha de tendência")

plt.xlabel(col1)
plt.ylabel(col2)
plt.title(f"Dispersão {col1} vs {col2} (r={r:.2f})")
plt.legend()

plt.savefig("grafico_dispersao.png", dpi=300, bbox_inches="tight")
plt.show()
```
